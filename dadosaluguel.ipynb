# Análises Exploratória

 Analisar quais bairros possuem a média de valor de aluguel mais elevadas

```python
df_aluguel_elevados = dados.groupby('Bairro')['Valor'].mean().sort_values(ascending=False).to_frame()
df_aluguel_elevados.rename(columns={'Valor': 'Percentuais'}, inplace=True)
df_aluguel_elevados.head(10)
```

Criar um gráfico de barras horizontais que apresente os 5 bairros com as médias de valores de aluguel mais elevadas

```python
df_top5_aluguel = df_aluguel_elevados.head(5)
df_top5_aluguel.plot(kind='bar', figsize=(14,10), color='blue',xlabel='Bairros',ylabel='Média de Aluguel',title='Top 5 Bairros com Média de Aluguel Mais Elevadas')
```

# Verificar si tem dados Nulos e Tratar

```python
df_apartamentos.isnull().sum()
```

Os valores nulos são iguias a True, somando todos os True = 1

Reemplazar esos valores Nulos por Zero

```python
dados_atualizados = df_apartamentos.fillna(0)
dados_atualizados.isnull().sum()
```

Removendo filas dos registros nulos

```python
registros_a_remover = dados_atualizados.query('Valor == 0 | Condominio == 0').index
dados_atualizados.drop(registros_a_remover,axis=0,inplace = True)
dados_atualizados.isnull().sum()
```

```python
dados_atualizados.query('Valor == 0 | Condominio == 0')
```

```python
dados_atualizados.head()
```

```python
dados_atualizados.Tipo.unique()
```

```python
dados_atualizados.drop('Tipo',axis=1,inplace=True)
dados_atualizados.head()
```

#Apartamentos que possuem 1 quarto e aluguel menor de 1200

```python
condicao1 = dados_atualizados['Quartos']==1
dados_atualizados[condicao1]
```

```python
condicao2 = dados_atualizados['Valor'] < 1200
dados_atualizados[condicao2]
```

```python
dados_1quarto_alueguel = dados_atualizados[condicao1 & condicao2]
```

#Apartamentos que possuem pelo menos 2 quartos, aluguel menor que 3000 e area maior que 70

```python
condicoes = (dados_atualizados['Quartos'] >= 2) & (dados_atualizados['Valor'] < 3000) & (dados_atualizados['Area'] > 70)
dados_condicoes_varias = dados_atualizados[condicoes]
dados_condicoes_varias
```

# Salvando os dados, no mesmo formato que foi entregado, seprado por ; os campos do arquivo CSV

```python
dados_atualizados.to_csv('dados_apartamentos.csv', index=False,sep=';')
dados_1quarto_alueguel.to_csv('dados_apartamentos_1quarto.csv', index=False,sep=';')
dados_condicoes_varias.to_csv('dados_apartamentos_condicionesvarias.csv', index=False,sep=';')
```

# Criando colunas numéricas

```python
dados.head()
```

Criando uma coluna valor por mês

```python
dados['Valor_por_mes'] = dados['Valor'] + dados['Condominio']
dados.head()
```

Criando uma coluna valor por ano

```python
dados['Valor_por_ano'] = (dados['Valor_por_mes'] *12) + dados['IPTU']
dados.head()
```

# Criando colunas categóricas

Cirando coluna Descrição

```python
dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro']
dados.head()
```

Criando colunas que possuim suite
  

```python
dados['Possui_Suite'] = dados['Suites'].apply(lambda x: 'Sim' if x > 0 else 'Não')
dados.head()
```

# Salvando Ddados con novas colunas

```python
dados.to_csv('dados_completos_dev.csv', index=False,sep=';')
```
